```python
import numpy as np
import matplotlib.pyplot as plt
import adi 

center_freq = 433.9e6 # of thermometer
num_samps=10000 #place holder value do we need this many samples?
sdr = adi.Pluto("ip:192.168.2.1") #class in adi.ad936x device 
sdr.sample_rate= int(2e6) # sample rate = 2MHz way more than the 3KHz bandwidth wireless signal.
sdr.rx_lo= int(center_freq) # carrier freq of Rx
sdr.gain_control_mode_chan0 = "manual" # can set gain of Rx and turns off AGC which changes the gain over time?
sdr.rx_hardwaregain_chan0 = 70.0 #dB Placeholder value change if needed?
sdr.rx_rf_bandwidth=int(2e6)
sdr.rx_buffer_size = num_samps # number of samples returned 


samples=sdr.rx() #receive samples off Pluto 

def spectrogram_plot(y,Fs,N):
    plt.specgram(y,NFFT=N,Fs= sdr.sample_rate,noverlap=N//2) #matplotlib syntax for spectrogram
    plt.ylabel("Frequency [Hz]")
    plt.xlabel("Time [S]")
    plt.colorbar(label="Power [dB]") #labels colorbar 
    plt.show()


spectrogram_plot(samples,sdr.sample_rate,1024)
```
